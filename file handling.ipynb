```python
import json
```

```python
fruits={'apple':100,'banana':200,'carrot':300}

with open('kumar.json','w')as f:
  json.dump(fruits,f)
```

```python
with open('kumar.json','r')as f:
  result=json.load(f)
print(result)
```

```python
fruits={'appple':100,'banana':200,'carrot':300}
print(fruits)
print(type(fruits))

res=json.dumps(fruits)
print(res)
print(type(res))
```

```python
fruits={'appple':100,'banana':200,'carrot':300}
print(fruits)
print(type(fruits))

res=json.dumps(fruits)
print(res)
print(type(res))

final=json.loads(res)
print(final)
print(type(final))
```
