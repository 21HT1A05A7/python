```python
class PRIME:
  def __init__(self,a,b):
    self.a=a
    self.b=b
  def fun(self):
     for i in range(self.a,self.b):
        c=0
        for j in range(1,i+1):
           if i%j==0:
             c=c+1
        if c==2:
          print(f'prime :{i}')
        else:
          print(f'not prime :{i}')
anil=PRIME(1,101)
anil.fun()
```
